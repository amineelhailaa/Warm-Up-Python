```python
1. Objectif du projet

Développer une application en ligne de commande (CLI) permettant de gérer un répertoire de contacts. Ce projet mobilise uniquement les notions abordées durant les Jours 1 à 4 : les variables, les conditions, les boucles, les fonctions, les listes, les dictionnaires, la gestion des fichiers et les exceptions.

2. User Story

En tant qu'utilisateur, je souhaite disposer d'un outil en ligne de commande pour gérer mon répertoire. Je veux pouvoir ajouter un contact (avec ses informations), consulter la liste complète, supprimer un contact, et m'assurer que mes données sont sauvegardées dans un fichier contacts.txt pour les retrouver à chaque ouverture du programme.

3. Environnement technique

    Langage : Python 3.x.
    Environnement : Éditeur local (VS Code) avec exécution dans le terminal.
    Contraintes : Utilisation exclusive des bibliothèques natives de Python.

4. Architecture des fonctionnalités

Partie 1 : Ajouter un contact

    L’utilisateur saisit le nom, le téléphone et l'email du contact.
    Le contact est enregistré dans une liste (en mémoire) et immédiatement sauvegardé dans contacts.txt.

Partie 2 : Afficher les contacts

    Le programme lit le fichier contacts.txt.
    Il affiche tous les contacts enregistrés, numérotés pour permettre une sélection facile.

Partie 3 : Supprimer un contact

    L’utilisateur choisit le numéro d'un contact dans la liste.
    Le programme retire ce contact de la liste et réécrit le fichier contacts.txt pour mettre à jour la base de données.

Partie 4 : Sauvegarde dans un fichier texte

    Utiliser un fichier contacts.txt où chaque ligne représente un contact.
    À chaque ajout ou suppression, le fichier est réécrit pour garantir la persistance des données.
```

```python
def deleteContact( contactNum ):
    with open("contacts.txt", "r", encoding= "utf-8") as file:
        contacts = file.readlines();
    for number,contact in enumerate(contacts, start=1):
        if contactNum == number:
            contacts.remove(contact)
            break

    with open("contacts.txt", "w", encoding="utf-8") as file:
        file.writelines(contacts) 
    print ("contact deleted successfully")
```

```python
deleteContact(1)
```
